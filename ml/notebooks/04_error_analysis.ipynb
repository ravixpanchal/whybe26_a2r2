# Phase 5 — error analysis

This view uses the persisted row-level predictions generated by the evaluation script. Errors are descriptive and do not tune the test set.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / "processed" / "train.csv").exists(): ROOT = ROOT.parent
predictions = pd.read_csv(ROOT / "ml" / "reports" / "predictions.csv")
summary = predictions.groupby(["split", "model"]).apply(lambda d: pd.Series({"rows": len(d), "false_positives": ((d.y_true == 0) & (d.prediction == 1)).sum(), "false_negatives": ((d.y_true == 1) & (d.prediction == 0)).sum()}), include_groups=False).reset_index()
summary
```
